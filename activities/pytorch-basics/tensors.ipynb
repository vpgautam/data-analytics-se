## Tensors

###  Introduction
In PyTorch, tensors are used to encode input, output and parameters of the model. Tensors are similar to ndarrays in numpy, except that they can gpus and other specialized hardware to acclerate computing.

```python
import torch
import numpy as np
```

### **Initialization**

#### **1. Directly from Data**

```python
data = [[1, 2], [3, 4]]
tensor_data = torch.tensor(data)
tensor_data_size = tensor_data.size()
print(f'tensor data = {tensor_data}, it has a size of {tensor_data_size}')
```

#### **2. From a NumPy array**

```python
np_array = np.array(data)
x_np = torch.from_numpy(np_array)
```

#### **3. From another Tensor**

The new tensor retains the properties (shape, datatype) of the argument tensor, unless explicitely overridden

```python
tensor_ones = torch.ones_like(tensor_data) # retains the properties of x_data
print(f'Ones Tensor: \n {tensor_ones} \n')
print(f'The datatype of ones tensor is {tensor_ones.dtype}')

tensor_rand = torch.rand_like(tensor_data, dtype=torch.float) # Overrides the datatype of x_data
print(f"Random Tensor: \n {tensor_rand} \n")
print(f"The datatype of tensor rand is  {tensor_rand.dtype}")
```

#### **4. With a Random or Constant values**

`shape` is a tuple of tensor dimensions. In the functions below, it determines the dimensionality o the output tensor.

```python
shape = (2, 3)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print(f"Random Tensor: \n {rand_tensor} \n")
print(f"Ones Tensor: \n {ones_tensor} \n")
print(f"Zeros Tensor: \n {zeros_tensor} ")
```

### **Tensor Attributes**

Tensor attributes describe tehir shape, datatype, and the device on which they are stored.

```python
tensor_data = torch.rand(3, 4)

print(f"The shape of tensor is {tensor_data.shape}.")
print(f"The datatype of tensor is {tensor_data.dtype}.")
print(f"Tensor data is stored in  {tensor_data.device}.")
```

### **Tensor Operations**
Many operations that are part of numpy array are defined for tensors. Some of the most important operations and concepts are described below.

Tensors are suited for running the algorithm on GPUs. 

```python
# We move our tensor to the GPU if available
device = torch.accelerator.current_accelerator().type if torch.accelerator.is_available() else 'cpu'
tensor_data = tensor_data.to(device)
print(f"Device tensor is stored on: {tensor_data.device}")
```

##### **Standard numpy-like indexing and slicing:**

```python
tensor = torch.ones(4, 4)
tensor[:,1] = 0
print(tensor)
print(f"\n Tensor is stored in {tensor.device}")
```

*Hence, data is usually stored in cpu, unless it is moved to gpu.*
**Joining tensors**: `torch.cat` , `torch.stack`

```python
t1 = torch.cat([tensor, tensor, tensor], dim=1)
print(t1)
```

**Element-wise Multplication:** `tensor.mul(tensor)`, `tensor * tensor`


**Matrix Multiplication:** `tensor.matmul(tensor.T)`, `tensor @ tensor.T`

**In-place operations:** In-place operations in PyTorch are operations that directly modify the content of an existing tensor instead of creating a new tensor with the result, thereby improving memory efficiency. These operations have an `_` at the end of function name. `tensor.add_(5)`, ``tensor.copy(y)``, `tensor.t()`

> In-place operations can break the computational graph because they modify the tensor in-place, which may lead to incorrect gradient calculations.

### Bridging with NumPy
Tensors on the CPU and NumPy arrays can share their underlying memory locations, and changing one will change the other

#### Tensor to NumPy array

```python
t = torch.ones(5)
print(f"t: {t}")
n = t.numpy()
print(f"n: {n}")
```

A change in the tensor reflects in the NumPy array.

```python
t.add_(1)
print(f"t: {t}")
print(f"n: {n}")
```

#### NumPy array to Tensor

```python
n = np.ones(5)
t =torch.from_numpy(n)
```

Changes in the NumPy array reflects in the tensor.

```python
np.add(n, 1, out=n)
print(f"t: {t}")
print(f"n: {n}")
```
